```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Step 1: Import the dataset
df = pd.read_csv("C:\\Users\\landg\\Downloads\\City_Air_Quality.csv", parse_dates=["Date"])
```

```python
# Step 2: Explore the dataset
print(df.head())
print(df.describe())
```

```python
# Step 3: Line plot for AQI trend over time
plt.figure(figsize=(10, 5))
plt.plot(df["Date"], df["AQI"], color='green', marker='o')
plt.title("AQI Trend Over Time")
plt.xlabel("Date")
plt.ylabel("AQI")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Step 4: Line plots for PM2.5, PM10, CO trends
plt.figure(figsize=(12, 6))
plt.plot(df["Date"], df["PM2.5"], label='PM2.5', marker='o')
plt.plot(df["Date"], df["PM10"], label='PM10', marker='x')
plt.plot(df["Date"], df["CO"], label='CO', marker='s')
plt.title("Pollutant Levels Over Time")
plt.xlabel("Date")
plt.ylabel("Concentration")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Step 5: Bar plot for AQI values by date
plt.figure(figsize=(10, 5))
plt.bar(df["Date"].dt.strftime('%Y-%m-%d'), df["AQI"], color='purple')
plt.title("AQI by Date")
plt.xlabel("Date")
plt.ylabel("AQI")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Step 6: Box plot for AQI
plt.figure(figsize=(6, 4))
plt.boxplot(df["AQI"])
plt.title("Box Plot of AQI Values")
plt.ylabel("AQI")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Step 7: Scatter plot between AQI and Pollutants
plt.figure(figsize=(10, 6))
plt.scatter(df["PM2.5"], df["AQI"], color='red', label="PM2.5 vs AQI")
plt.scatter(df["PM10"], df["AQI"], color='blue', label="PM10 vs AQI")
plt.scatter(df["CO"], df["AQI"], color='green', label="CO vs AQI")
plt.title("Scatter Plot of AQI vs Pollutants")
plt.xlabel("Pollutant Level")
plt.ylabel("AQI")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
