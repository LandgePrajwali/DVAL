```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
```

```python
# Step 1: Import the dataset
df = pd.read_csv("C:\\Users\\landg\\Downloads\\RealEstate_Prices.csv")
```

```python
# Step 2: Clean column names
df.columns = (
    df.columns.str.strip()
    .str.replace(r"[^\w\s]", "", regex=True)
    .str.replace(" ", "_")
    .str.lower()
)
```

```python
# Step 3: Handle missing values
df["sale_price"].fillna(df["sale_price"].mean(), inplace=True)
```

```python
# Step 4: Example merge (optional - here we simulate a merge)
demographics = pd.DataFrame({
    "location": ["Central City", "Uptown", "Suburb"],
    "avg_income": [70000, 85000, 60000]
})
df = df.merge(demographics, how="left", on="location")
```

```python
# Step 5: Filter data (e.g., only "House" properties)
filtered_df = df[df["property_type"] == "House"]
```

```python
# Step 6: Encode categorical variables
df_encoded = pd.get_dummies(df, columns=["location", "property_type"], drop_first=True)
```

```python
# Step 7: Aggregate data (e.g., avg sale price by property type)
agg = df.groupby("property_type")["sale_price"].mean().reset_index(name="avg_price")
```

```python
# Step 8: Handle outliers using IQR method
Q1 = df["sale_price"].quantile(0.25)
Q3 = df["sale_price"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df_cleaned = df[(df["sale_price"] >= lower_bound) & (df["sale_price"] <= upper_bound)]
```

```python
# Save cleaned dataset
df_cleaned.to_csv("Cleaned_RealEstate_Prices.csv", index=False)
print("Cleaned dataset saved.")
```
