```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Task 1: Import the dataset
df = pd.read_csv("C:\\Users\\landg\\Downloads\\Housing_Prices.csv.csv")
```

```python
# Task 2: Explore the dataset
print("First 5 rows:\n", df.head())
print("\nData Types:\n", df.dtypes)
print("\nSummary:\n", df.describe())
```

```python
# Task 3: Convert relevant columns to NumPy ndarray
prices = df["Price"].to_numpy()
sizes = df["Size_sqft"].to_numpy()
```

```python
# Task 4: NumPy statistical operations
print("\nMean Price:", np.mean(prices))
print("Median Price:", np.median(prices))
print("Standard Deviation:", np.std(prices))
print("Price Range:", np.max(prices) - np.min(prices))
```

```python
# Task 5: Data cleaning (example: checking for missing values)
print("\nMissing Values:\n", df.isnull().sum())
```

```python
# Task 6: Filtering and subsetting
# Houses with more than 3 bedrooms
print("\nHouses with more than 3 bedrooms:\n", df[df["Bedrooms"] > 3])
```

```python
# Houses priced between 400,000 and 500,000
print("\nHouses priced between 400,000 and 500,000:\n", df[(df["Price"] >= 400000) & (df["Price"] <= 500000)])
```

```python
# Task 7: Descriptive statistics
print("\nPrice Percentiles:\n", df["Price"].quantile([0.25, 0.5, 0.75]))
print("\nFrequency Table for Bedrooms:\n", df["Bedrooms"].value_counts())
```

```python
# Task 8: Aggregation
print("\nAverage Price by Location:\n", df.groupby("Location")["Price"].mean())
print("\nAverage Price by Bedrooms:\n", df.groupby("Bedrooms")["Price"].mean())
```

```python
# Task 9: (Optional) Visualization
df["Price"].plot(kind="hist", bins=5, title="Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```
